```python
import requests

base_url = "http://mtg.upf.edu/static/datasets/last.fm"
file_name = "lastfm-dataset-1K.tar.gz"
download_url = f"{base_url}/{file_name}"
output_path = f"../data"
output_file = f"{output_path}/{file_name}"
```

```python
print("Downloading file...")
with requests.get(download_url, stream=True) as response:
    response.raise_for_status()
    final_file_size_in_bytes = int(response.headers.get("Content-Length"))
    total_file_size_in_mb = None
    if final_file_size_in_bytes is None:
        print("Total File size is unknown. 'Content-Length' header is missing in response")
    else:
        total_file_size_in_mb = round(final_file_size_in_bytes / 1000 / 1000)
        print(f"Total File size: {str(total_file_size_in_mb)} MB")
        
    with open(output_file, 'wb') as file:
        chunk_size = 1024 * 1024
        for i, chunk in enumerate(response.iter_content(chunk_size=chunk_size)):
            file.write(chunk)
            already_downloaded_in_mb = round(i * chunk_size / 1000 / 1000)
            if i * chunk_size / 1000 / 1000 == 0 or already_downloaded_in_mb % 50 == 0:
                print(f"Progress: {str(already_downloaded_in_mb)} MB / {str(total_file_size_in_mb) if total_file_size_in_mb is not None else 'unknown'} MB")
print(f"Successfully downloaded the file to {output_file}")
```

```python
import tarfile

with tarfile.open(output_file, "r:gz") as tar:
    print(f"Extracting all files from {output_file}")
    tar.extractall(path=output_path)
    print(f"Successfully extracted all files into {output_path}")
```
